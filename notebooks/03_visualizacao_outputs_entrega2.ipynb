# Visualizacao dos Outputs da Entrega 2 Revisada

Este notebook serve para gerar evidencias visuais da base antes e depois das transformacoes pedidas na reabertura da Entrega 2.

Ele le os arquivos gerados em `outputs/`, mostra amostras com `head(20)` e exibe os parametros salvos no metadata, principalmente os parametros da normalizacao.


## 1. Carregamento dos Arquivos


```python
from pathlib import Path
import json

import pandas as pd

pd.set_option('display.max_columns', 120)
pd.set_option('display.width', 160)
pd.set_option('display.max_colwidth', 80)

current_dir = Path.cwd()
project_dir = current_dir.parent if current_dir.name == 'notebooks' else current_dir
output_dir = project_dir / 'outputs'

cleaned_path = output_dir / 'drug_events_cleaned.csv'
train_path = output_dir / 'drug_events_train_transformed.csv'
test_path = output_dir / 'drug_events_test_transformed.csv'
sample_path = output_dir / 'drug_events_sample.csv'
metadata_path = output_dir / 'data_preparation_metadata.json'

cleaned_df = pd.read_csv(cleaned_path)
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)
sample_df = pd.read_csv(sample_path)

with metadata_path.open(encoding='utf-8') as f:
    metadata = json.load(f)

print('Arquivos carregados:')
print(f'  Antes / base limpa: {cleaned_df.shape} -> {cleaned_path.name}')
print(f'  Depois / treino transformado: {train_df.shape} -> {train_path.name}')
print(f'  Depois / teste transformado: {test_df.shape} -> {test_path.name}')
print(f'  Amostra: {sample_df.shape} -> {sample_path.name}')
print(f'  Metadata: {metadata_path.name}')
```

## 7. Resumo dos Dados Gerados


```python
summary = pd.DataFrame([
    {
        'arquivo': 'drug_events_cleaned.csv',
        'papel': 'base limpa antes da nova transformacao',
        'registros': len(cleaned_df),
        'colunas': cleaned_df.shape[1],
    },
    {
        'arquivo': 'drug_events_train_transformed.csv',
        'papel': 'treino depois da transformacao',
        'registros': len(train_df),
        'colunas': train_df.shape[1],
    },
    {
        'arquivo': 'drug_events_test_transformed.csv',
        'papel': 'teste depois da transformacao',
        'registros': len(test_df),
        'colunas': test_df.shape[1],
    },
    {
        'arquivo': 'drug_events_sample.csv',
        'papel': 'amostra visual para conferencia',
        'registros': len(sample_df),
        'colunas': sample_df.shape[1],
    },
])
summary
```

## 8. Metadata: Parametros Salvos

A professora pediu que, na normalizacao, o conjunto de treino fosse normalizado, os parametros fossem gravados e depois aplicados no teste.

O arquivo `data_preparation_metadata.json` e justamente a evidencia disso: ele registra as medias e desvios aprendidos no treino, alem de informar que o KNN e o StandardScaler foram ajustados apenas com dados de treinamento.


```python
scaler_params = metadata['scaler_parameters']
scaler_df = pd.DataFrame({
    'feature': scaler_params['features'],
    'media_aprendida_no_treino': scaler_params['mean'],
    'desvio_aprendido_no_treino': scaler_params['scale'],
})
scaler_df
```

## 9. Metadata: Imputacao KNN e Split


```python
knn_info = metadata['age_group_imputation']
knn_summary = pd.DataFrame([
    {'item': 'metodo', 'valor': knn_info['method']},
    {'item': 'escopo do ajuste', 'valor': knn_info['fit_scope']},
    {'item': 'alvo imputado', 'valor': knn_info['target']},
    {'item': 'registros imputados no treino', 'valor': knn_info['train_imputed_records']},
    {'item': 'registros imputados no teste', 'valor': knn_info['test_imputed_records']},
])
knn_summary
```

## 10. Distribuicao da Variavel Alvo


```python
target_distribution = pd.DataFrame({
    'treino': train_df['serious'].value_counts().sort_index(),
    'teste': test_df['serious'].value_counts().sort_index(),
})
target_distribution['treino_%'] = (target_distribution['treino'] / len(train_df) * 100).round(2)
target_distribution['teste_%'] = (target_distribution['teste'] / len(test_df) * 100).round(2)
target_distribution
```

## 11. Conferencia da Normalizacao


```python
normalized_features = metadata['dataset_info']['numeric_features_normalized']
normalization_check = pd.concat(
    {
        'treino_media': train_df[normalized_features].mean(),
        'treino_desvio': train_df[normalized_features].std(),
        'teste_media': test_df[normalized_features].mean(),
        'teste_desvio': test_df[normalized_features].std(),
    },
    axis=1,
)
normalization_check
```
